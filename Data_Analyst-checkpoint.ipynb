```python
# 1. Clean the index directly (since it's already set)
df.index = df.index.astype(str).str.upper().str.strip()

# 2. Filter the rows based on the index values
# This keeps all 17 columns and only drops the 'NAN' or empty rows
df = df[df.index.notna() & ~df.index.isin(['NAN', ''])]

# Verification
print(f"Data successfully cleaned! New shape: {df.shape}")
df
```

```python
import pandas as pd

# 1. Define your columns by category
text_cols = [
    'job_title', 'job_location', 'job_via', 'job_schedule_type', 
    'search_location', 'job_country', 'company_name', 'job_skills', 'job_type_skill'
]

numeric_cols = ['salary_rate', 'salary_year_avg', 'salary_hour_avg']

boolean_cols = ['job_work_from_home', 'job_no_degree_mention', 'job_health_insurance']

date_cols = ['job_posted_date']

# 2. Clean Text Columns (Uppercase, strip whitespace, handle 'NAN' strings safely)
for col in text_cols:
    if col in df.columns:
        # Standardize text and strip spaces
        df[col] = df[col].astype(str).str.upper().str.strip()
        # Convert literal 'NAN' or 'NONE' strings back to actual NaN values
        df[col] = df[col].replace(['NAN', 'NONE', ''], pd.NA)

# 3. Clean Numeric Columns (Ensure they stay numeric, don't convert to string)
for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# 4 Define your 3 boolean columns
boolean_cols = ['job_work_from_home', 'job_no_degree_mention', 'job_health_insurance']

for col in boolean_cols:
    if col in df.columns:
        # 1. Convert to string, uppercase, and strip to clean up any messy text formatting
        clean_text = df[col].astype(str).str.upper().str.strip()
        
        # 2. Map the text values to actual Python Booleans (True / False)
        # Anything else (like 'NAN', 'NONE', or empty spaces) becomes a proper missing value (None)
        df[col] = clean_text.map({'TRUE': True, 'FALSE': False}).astype('boolean')

# 5. Clean Date Column (Convert to proper datetime objects)
for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')

df = df.set_index('job_title_short')

# Check your final clean DataFrame
print(df.info())
df
```

```python
# 1. Isolate the dictionary column alongside core variables
sub_types = df_africa[['job_title_short', 'job_country', 'salary_year_avg', 'job_type_skills']].dropna(subset=['job_type_skills']).copy()

# 2. Extract nested skills from dictionary values
def extract_dict_skills(val):
    if pd.isna(val):
        return []
    try:
        # Converts the raw text response into a real Python dictionary
        parsed_dict = ast.literal_eval(str(val).strip())
        if isinstance(parsed_dict, dict):
            all_skills = []
            # Loop through all keys ('analyst_tools', 'programming', etc.) and unpack their lists
            for skill_list in parsed_dict.values():
                if isinstance(skill_list, list):
                    all_skills.extend(skill_list)
            return [str(s).upper().strip() for s in all_skills]
    except:
        return []
    return []

# Run the parser on your multiple-response column
sub_types['job_type_clean'] = sub_types['job_type_skills'].apply(extract_dict_skills)

# 3. Explode ONLY the flat list column to prevent crashing
exploded_types = sub_types.explode('job_type_clean').dropna(subset=['job_type_clean'])

# 4. Generate the Cross-Analysis Table
analysis_col2 = exploded_types.groupby(['job_title_short', 'job_country', 'job_type_clean']).agg(
    skill_count=('job_type_clean', 'count'),
    mean_salary=('salary_year_avg', 'mean')
).reset_index()

# 5. Sort the final output layout
analysis_col2 = analysis_col2.sort_values(by=['job_country', 'skill_count'], ascending=[True, False])

print("--- Column 2 (job_type_skills) Analysis Successfully Completed ---")
print(analysis_col2.head(10))
```
