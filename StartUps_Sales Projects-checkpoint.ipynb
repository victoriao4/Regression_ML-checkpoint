```python
import pandas as pd
import numpy as np

# ========= 1. CONFIG =========
POP_IND = 8_000_000
durations = ["10Minutes","15Minutes","30Minutes","60Minutes","MonthlyFee"]
test_prices = {"10Minutes":5000, "15Minutes":10000, "30Minutes":15000, "60Minutes":20000, "MonthlyFee":1000}

DISP_COLS = {
    "Individual": {
        "10Minutes": "Disposition10MinutesN5,000",
        "15Minutes": "Disposition15MinutesN10,000", 
        "30Minutes": "Disposition30MinutesN15,000",
        "60Minutes": "Disposition60MinutesN20,000",
        "MonthlyFee": "Disposition1MonthlyFeeN1,000" # <--- EXACT NAME FROM YOUR SCREENSHOT
    }
}


# ========= 2. HELPER FUNCTIONS =========
def get_top2box_pct(df, col):
    """Calculate % Extremely + Very Disposed"""
    disp = df[col].astype(str)
    buy = disp.isin(['Extremely Disposed', 'Very Disposed', '1', '5'])
    return buy.mean()


def leakage_all5_fixed(df, POP, summary_df):
    """
    Runs VW Acceptance model for all 5 concepts
    Disposition at OPP = 1 if OPP falls within respondent's [TooCheap, TooExpensive] range
    """
    n = len(df)
    results = []
    
    print(f"Running N={n} | Scaling to POP={POP/1e6:.1f}M\n")
    
    for dur in durations:
        prop_price = test_prices[dur]
        opp = summary_df.loc[summary_df['Concept']==dur, 'OPP'].values[0]
        disp_col = DISP_COLS["Individual"][dur]
        
        # SAFETY CHECK
        if disp_col not in df.columns:
            print(f"SKIPPING {dur}: Column '{disp_col}' not found")
            continue
        
        # 1. OBSERVED DISPOSITION AT PROPOSED PRICE
        buy_pct_prop = get_top2box_pct(df, disp_col)
        
        # 2. MODELED DISPOSITION AT OPP USING VW
        tc_col = f'PriceTooCheapInadequate{dur}'
        c_col = f'PriceCheapadequate{dur}'
        e_col = f'PriceExpensiveadequate{dur}'
        te_col = f'PriceTooExpensiveNeverBuy{dur}'
        
        # MonthlyFee has different naming in VW
        if dur == 'MonthlyFee':
            e_col = 'PriceExpensiveadequateMonthlyFee'
            te_col = 'PriceTooExpensivNeverBuyMonthlyFee'
        
        tc = pd.to_numeric(df[tc_col], errors='coerce')
        c = pd.to_numeric(df[c_col], errors='coerce')
        e = pd.to_numeric(df[e_col], errors='coerce')
        te = pd.to_numeric(df[te_col], errors='coerce')
        
        price_min = pd.concat([tc, te], axis=1).max(axis=1) # lowest they accept
        price_max = pd.concat([c, e], axis=1).min(axis=1) # highest they accept
        accepts_opp = ((price_min <= opp) & (price_max >= opp))
        buy_pct_opp = accepts_opp.mean()
        
        # 3. PROJECT TO POPULATION
        vol_prop = buy_pct_prop * POP
        vol_opp = buy_pct_opp * POP
        rev_prop = vol_prop * prop_price
        rev_opp = vol_opp * opp
        leakage = rev_opp - rev_prop # negative = gain vs OPP
        
        results.append({
            'Concept': dur,
            'Prop Price': f"N{prop_price:,}",
            'OPP': f"N{round(opp):,}",
            'Buy% Prop': f"{buy_pct_prop*100:.1f}%",
            'Buy% OPP': f"{buy_pct_opp*100:.1f}%",
            'Vol Prop M': f"{vol_prop/1e6:.2f}M",
            'Vol OPP M': f"{vol_opp/1e6:.2f}M",
            'Rev Prop Bn': f"N{rev_prop/1e9:.2f}B",
            'Rev OPP Bn': f"N{rev_opp/1e9:.2f}B",
            'Rev Gain vs OPP Bn': f"N{-leakage/1e9:.2f}B"
        })
        print(f"{dur}: Prop {buy_pct_prop*100:.1f}% @N{prop_price:,} vs OPP {buy_pct_opp*100:.1f}% @N{opp:,.0f} | Gain: N{-leakage/1e9:.2f}B")
    
    return pd.DataFrame(results)


# ========= 3. RUN =========
df_final_ind = leakage_all5_fixed(ind, POP_IND, df_ind_vw)

print("\n======== FINAL TABLE FOR DECK: INDIVIDUAL 8M ========")
display(df_final_ind)
```
